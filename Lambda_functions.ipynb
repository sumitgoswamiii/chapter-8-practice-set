```python
# Method 2
c = []
def cube(a):
  for elem in a:
    c.append(elem**3)
  return c
cube([1,2,3,4,5,6])
```

```python
# Write a code to return a list with the plus 5 of every number in the list.
def sum(lst):
  cb = []
  for elem in lst:
    cb.append(elem + 5)
  return cb
sum([1, 2, 3, 4, 5, 6, 7])
```

```python
# Given a list of strings, convert all the strings to all uppercase and display the list
def uppercase(lst):
  ups = []
  for string in lst:
    ups.append(string.upper())
  return ups
uppercase(["sumit", "rohan", "vikash", "aditya"])
```

```python
uppercase = list(map(lambda x: x.upper(), ["sumit", "vikash", "rohan","aditya"]))
print(uppercase)
```

```python
# Given a list of prices, return a new list with 18% GST included prices.
def GST(l1):
  new_l1 = []
  for elem in l1:
    new_l1.append(elem + elem*(18/100))
  return new_l1
GST([100, 200, 250, 350, 700 ])
```

```python

new_lst = list(map(lambda x: x + x*0.18, [100, 200, 250, 350, 700 ]))
print(new_lst)
```

# Filter

```python
num = [1, 2, 3, 4, 5, 6, 7]
#Write a code to create a list of all even numbers of the given list
```

```python
def even(num):
  new_ev = []
  for elem in num:
    if elem % 2 == 0:
      new_ev.append(elem)
  return new_ev
even(num)
```

```python
new_even = list(filter(lambda x:x % 2 == 0 , [1, 2, 3, 4, 5, 6, 7]))
print(new_even)
```
